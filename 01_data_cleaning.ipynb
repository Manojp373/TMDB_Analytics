```python
import pandas as pd
import numpy as np

movies = pd.read_csv("DATA\movies.csv")
genres = pd.read_csv("DATA\genres.csv")
cast = pd.read_csv("DATA\cast.csv")
crew = pd.read_csv("DATA\crew.csv")
movie_keywords = pd.read_csv("DATA\movie_keywords.csv")
movie_genres = pd.read_csv("DATA\movie_genres.csv")
```

```python
datasets = {
    "movies": movies,
    "genres": genres,
    "cast": cast,
    "crew": crew,
    "movie_keywords": movie_keywords,
    "movie_genres": movie_genres
}

for name, df in datasets.items():
    print(name, df.shape)
```

```python
movies['budget'] = movies['budget'].replace(0, np.nan)

movies['revenue'] = movies['revenue'].replace(0, np.nan)
```

```python
budget_median = movies['budget'].median()

revenue_median = movies['revenue'].median()

print("Budget Median:", budget_median)
print("Revenue Median:", revenue_median)
```

```python
movies['budget'] = movies['budget'].fillna(budget_median)

movies['revenue'] = movies['revenue'].fillna(revenue_median)
```

```python
print("Zero budgets:", (movies['budget'] == 0).sum())
print("Zero revenues:", (movies['revenue'] == 0).sum())

print("Missing budgets:", movies['budget'].isnull().sum())
print("Missing revenues:", movies['revenue'].isnull().sum())
```

```python
placeholders = ['Unknown', 'N/A', 'None', 'null', 'Not Available', '']

for name, df in datasets.items():
    print("\nDataset:", name)

    for col in df.select_dtypes(include='object').columns:
        count = df[col].astype(str).str.strip().isin(placeholders).sum()

        if count > 0:
            print(col, ":", count)
```

```python
orphan_cast = cast[
    ~cast['movie_id'].isin(movies['movie_id'])
]

print("Orphaned cast records:", len(orphan_cast))
```

```python
orphan_crew = crew[
    ~crew['movie_id'].isin(movies['movie_id'])
]

print("Orphaned crew records:", len(orphan_crew))
```

```python
orphan_keywords = movie_keywords[
    ~movie_keywords['movie_id'].isin(movies['movie_id'])
]

print("Orphaned keyword records:", len(orphan_keywords))
```

```python
orphan_movie_genres = movie_genres[
    ~movie_genres['movie_id'].isin(movies['movie_id'])
]

print("Orphaned movie-genre records:", len(orphan_movie_genres))
```

```python
orphan_genre_ids = movie_genres[
    ~movie_genres['genre_id'].isin(genres['genre_id'])
]

print("Invalid genre references:", len(orphan_genre_ids))
```
