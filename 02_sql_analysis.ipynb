```python
mycursor.execute("""
CREATE TABLE IF NOT EXISTS movies (
    movie_id INT PRIMARY KEY,
    title VARCHAR(255),
    original_language VARCHAR(10),
    release_date DATE,
    runtime INT,
    budget BIGINT,
    revenue BIGINT,
    popularity FLOAT,
    vote_average FLOAT,
    vote_count INT,
    status VARCHAR(50),
    overview TEXT
)
""")

print("Movies table created!")
```

```python
mycursor.execute("""
CREATE TABLE IF NOT EXISTS genres (
    genre_id INT PRIMARY KEY,
    genre_name VARCHAR(100)
)
""")

print("Genres table created!")
```

```python
mycursor.execute("""
CREATE TABLE IF NOT EXISTS `cast` (
    cast_row_id INT PRIMARY KEY AUTO_INCREMENT,
    movie_id INT,
    person_id INT,
    actor_name VARCHAR(255),
    character_name TEXT,
    cast_order INT,
    FOREIGN KEY (movie_id) REFERENCES movies(movie_id)
)
""")

print("Cast table created!")
```

```python
mycursor.execute("""
CREATE TABLE IF NOT EXISTS crew (
    crew_row_id INT PRIMARY KEY AUTO_INCREMENT,
    movie_id INT,
    person_id INT,
    person_name VARCHAR(255),
    job VARCHAR(100),
    department VARCHAR(100),
    FOREIGN KEY (movie_id) REFERENCES movies(movie_id)
)
""")

print("Crew table created!")
```

```python
mycursor.execute("""
CREATE TABLE IF NOT EXISTS movie_genres (
    movie_id INT,
    genre_id INT,
    FOREIGN KEY (movie_id) REFERENCES movies(movie_id),
    FOREIGN KEY (genre_id) REFERENCES genres(genre_id)
)
""")

print("Movie Genres table created!")
```

```python
mycursor.execute("""
CREATE TABLE IF NOT EXISTS movie_keywords (
    movie_id INT,
    keyword_id INT,
    keyword_name VARCHAR(100),
    FOREIGN KEY (movie_id) REFERENCES movies(movie_id)
)
""")

print("Movie Keywords table created!")
```

```python
mycursor.execute("""SELECT genre_name, AVG(revenue),AVG(budget) FROM movies JOIN movie_genres USING (movie_id) JOIN genres USING (genre_id) GROUP BY genre_name ORDER BY AVG(revenue) DESC""")
out=mycursor.fetchall()
print(tabulate(out,headers=[i[0] for i in mycursor.description],  tablefmt='psql'))
```

```python
mycursor.execute("""SELECT actor_name, COUNT(movie_id) FROM cast GROUP BY actor_name ORDER BY COUNT(movie_id) DESC LIMIT 10""")
out=mycursor.fetchall()
print(tabulate(out,headers=[i[0] for i in mycursor.description],  tablefmt='psql'))
```

```python
mycursor.execute(""" SELECT person_name, COUNT(movie_id) FROM crew GROUP BY person_name HAVING COUNT(movie_id) > 3 ORDER BY COUNT(movie_id) DESC""")
out=mycursor.fetchall()
print(tabulate(out,headers=[i[0] for i in mycursor.description],  tablefmt='psql'))
```

```python
mycursor.execute("""SELECT keyword_name, COUNT(movie_id) FROM movie_keywords GROUP BY keyword_name ORDER BY COUNT(movie_id) DESC LIMIT 10""")
out=mycursor.fetchall()
print(tabulate(out,headers=[i[0] for i in mycursor.description],  tablefmt='psql'))
```
